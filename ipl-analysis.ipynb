```python
df['Runs_Scored'] = pd.to_numeric(df['Runs_Scored'], errors='coerce')
df['Sixes'] = pd.to_numeric(df['Sixes'], errors='coerce')
df['Wickets_Taken'] = pd.to_numeric(df['Wickets_Taken'], errors='coerce')

top_runs = df.groupby('Player_Name')['Runs_Scored'].sum().sort_values(ascending=False).head(10)
print("Top 10 Run Scorers:")
print(top_runs)
```

```python
import matplotlib.pyplot as plt
```

```python
top_sixes = df.groupby('Player_Name')['Sixes'].sum().sort_values(ascending=False).head(10)
print("Top 10 six Hitters:")
print(top_sixes)

plt.figure(figsize=(10,6))
plt.barh(top_sixes.index[::-1], top_sixes.values[::-1],color='#6B35FF')
plt.title('Top 10 IPL Six Hitters of All Time', fontsize=16)
plt.tight_layout()
plt.show()
```

```python
top_wickets = df.groupby('Player_Name')['Wickets_Taken'].sum().sort_values(ascending=False).head(10)
print("Top 10 Wicket Taken:")
print(top_wickets)

plt.figure(figsize=(10,6))
plt.barh(top_wickets.index[::-1], top_wickets.values[::-1],color='#35FF6B')
plt.title('Top 10 IPL Wickets TAkers of All Time', fontsize=16)
plt.xlabel('Total wickets', fontsize=12)
plt.tight_layout()
plt.show
```
