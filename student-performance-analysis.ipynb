```python
import zipfile
with zipfile.ZipFile("Student Performance in Exams.zip", "r") as z:
  z.extractall()
  print(z.namelist())
```

```python
import pandas as pd

df = pd.read_csv("StudentsPerformance.csv", encoding="latin1")
print(df.shape)
```

```python
df.isnull().sum()
```

```python
average_scores = df[['math score', 'reading score', 'writing score']].mean()
print(average_scores)
```

```python
average_scores = df['math score'].mean()
print(average_scores)
```

```python
parental_scores = df.groupby('test preparation course')[['math score', 'reading score','writing score']].mean()
print(parental_scores)
```

```python
df.groupby('parental level of education')[['math score', 'reading score','writing score']].mean()
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(10,6))
plt.barh(average_scores.index[::-1], average_scores.values[::-1], color='#2196F3')
plt.title('AVERAGE SCORES BY SUBJECT', fontsize=16)
plt.xlabel('AVERAGE SCORES', fontsize=12)
plt.tight_layout()
plt.show()
```

```python
average_scores = df[['math score', 'reading score', 'writing score']].mean()
print(average_scores)
```

```python
plt.figure(figsize=(10,6))
plt.barh(average_scores.index[::-1], average_scores.values[::-1], color='#2196F3')
plt.title('AVERAGE SCORES BY SUBJECT', fontsize=16)
plt.xlabel('AVERAGE SCORES', fontsize=12)
plt.tight_layout()
plt.show()
```

```python
parental_scores.plot(kind='bar', figsize=(8,5), color=['#2196F3','#4CAF50','#FF9800'])
plt.title('Test Prep vs No Prep Scores', fontsize=16)
plt.xlabel('Test Preparation', fontsize=12)
plt.ylabel('Average Score', fontsize=12)
plt.xticks(rotation=0)
plt.tight_layout()
plt.show
```

```python
parental_scores['math score'].plot(kind='bar', figsize=(10,6), color='#FF9800')
plt.title('Math Score by Parental Education Level', fontsize=16)
plt.xlabel('Parental Education', fontsize=12)
plt.ylabel('Average Score', fontsize=12)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
